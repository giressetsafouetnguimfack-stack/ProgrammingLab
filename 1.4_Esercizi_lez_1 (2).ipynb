### **1. Trasformare cicli in list comprehension**

Fai tre esempi di cicli producono una lista e poi riscrivi lo stesso codice usando la list comprehension

```python
#-1 Versione Con Cicli

risultat = []
for i in range(5):
    risultat.append(i*i)
print(risultat)

#-1 Versione Con List Comprehension

risultat = [i*i for i in range(5)]
print(risultat)


#-2 Versione Con Cicli

pari = []
for i in range(10):
    if i % 2 == 0:
        pari.append(i)
print(pari)  

#-2 Versione Con List Comprehension

pairi =[i for i in range(10) if i % 2 == 0]


#-3 Versione Con Cicli

words = ["mango", "orange", "banane"]
upper_words = []

for w in words:
    upper_words.append(w.upper())
print(upper_words)  

#-3 Version Con List Comprehension

upper_words = [w.upper() for w in ["mango", "orange", "banane"]]
print(upper_words)



 
```

### **2. Vettore di Numeri Primi**

1. **Crea un vettore** contenente tutti i numeri primi compresi tra **0 e 10**  
   _(Puoi scriverli direttamente nell'array)._

2. **Conta quanti numeri ci sono nel vettore** utilizzando la funzione `len()`.  **Ottieni lo stesso numero** accedendo all'attributo `.size` del vettore.

3. Quale pensi sia il tipo di dato (`dtype`) del vettore?  Prova a rispondere senza eseguire il codice e Verifica la tua risposta accedendo all'attributo `.dtype` del vettore

4. Scrivi l'array usando una list comprehension che controlla che i numeri siano primi. 

```python
#-1 Crea un vettore

import numpy as np

primes = []

for n in range(2, 11):
    primo = True 
    for i in range(2, n):
        if i % i == 0:
            primo = False
            break
    if primo:
        primes.append(n)  

prim = np.array(primes)    
print(prim)      


#-2 Conta quanti numeri sono...

import numpy as np

prim = np.array([2,3,5,7])

print(len(prim))

print(prim.size)


#-3 il tipo di dato

# il dato è  di tipo int

import numpy as np 

prim = np.array([2,3,5,7])
print(prim.dtype)

#-4 List comprehension

import numpy as np

prim = np.array([n for n in range(2, 11)
                 if all(n % i != 0 for i in range(2, n))])

print(prim)





```

### **3.  Operazioni su array**
Crea un array **a**, 1D (senza digitarlo esplicitamente), e genera due nuovi array **b**, contenente una sottostringa a piacere e **c** che è il reverse dell'array **a** (esempio [1,2,3] diventa [3,2,1]) .  Dividi l'array **a** per l'array **c**. Fai la stessa cosa per una lista.

```python
# Con un array

a = np.arange(1, 6)

b = a[1 : 4]

c = a[::-1]

risultat = a  / c

print("a =", a)
print("b =", b)
print("c =", c)
print("a / c =", risultat)


# Con Lista

a = list(range(1,6))

b = a[1:4]

c = [a::-1]

risultat = [a[i] / c[i] for i in range(len(a))]

print("a =", a)
print("b =", b)
print("c =", c)
print("a / c =", risultat)
```
